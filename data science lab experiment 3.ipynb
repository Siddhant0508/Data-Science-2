```python
#Normal Dist: mean=0
#curve= bell shaped curve
from numpy import random
#random function which is used to generate random data 
import matplotlib.pyplot as plt
#from matplotlib import pyplot only and as a plt alis
import seaborn as sns
```

```python
#Generate some random(Normalized)distribution of size 1x100
x=random.normal(size=100)
print(x)
```

```python
x=random.normal(size=(2,3))
print(x)
```

```python
#mean =1 and std=2
x=random.normal(loc=1,scale=2,size=(2,3))
#variable=random(function->random_data)-> normal(mean,std,size).
print(x)
```

```python
sns.displot(random.normal(size=1000),kind="kde")#Kernel Density Estimator 
#KDE is used to draw smooth curve over the data
plt.show()
```

```python
#random.binomial(n,p,size=None)
#N = no. of trials
#P = Probability 
#size= the size of the returned array
from numpy import random
x=random.binomial(n=10,p=0.5,size=10)
print(x)
```

```python
sns.displot(random.binomial(n=10,p=0.5,size=10),kde=False)
plt.show
```

POISSON DISTRIBUTION

```python
#IF ANT EVENT HAS ALREADY OCCURED FOR X TIME THEN WHAT IS TTHE PROBABILITY THAT THE SAME WILL HAPPEN AGAIN
#lam=Rate of known occurences of the event 
x=random.poisson(lam=2,size=7)
print(x)
```

```python
sns.displot(random.poisson(lam=2,size=1000),kde=False)
plt.show()
```

```python
# UNIFORM DISTRIBUTION
# every event has the equal chance of appearing 
# E.G : Generation of random number
x=random.uniform(size=(2,3))
print(x)
```

```python
sns.displot(random.uniform(size=1000),kde=True)
plt.show()
```

```python
# MULTINOMIAL DISTRIBUTION
# here any event has more than two possible outcomes 
x=random.multinomial(n=20,pvals=[1/6,1/6,1/6,1/6,1/6,1/6])
print(x)
# performing 20 dice rolls and each outcomes has 1/6 chance of appearing
# n = number of possible outcomes
# pval = list of probabilities of outcomes 
```

```python
sns.displot(random.multinomial(n=20,pvals=[1/6,1/6,1/6,1/6,1/6,1/6]))
plt.show
```

```python
# EXPONENTIAL DISTRIBUTION
from numpy import random
x=random.exponential(scale=2,size=(2,3))
print(x)
```

```python
sns.displot(random.exponential(size=1000),kde=True)
plt.show
```
